```python
# Set up file handler

TREE_NAME = "posteriors"
input_file = Path("/Users/henrywallace/T2K/AdaptiveTune/ForEdwin/trimmed_MaCh3-joint2021_asimovA22_uniformND280binning_woRC_v25Mar2022_Perlmutter_job1_1.root")
DATASET = ROOTDataset(INPUT_FILE, tree_name=TREE_NAME,
                      branches=["xsec*"],
                      labels=["LogL_systematic_xsec_cov"],
                      only_unique_entries=True)
```

```python
#Machine learning time

# Dimensions
INPUT_DIM = DATASET.get_n_features()
OUTPUT_DIM = DATASET.get_n_labels()
```

```python
# Layers

LAYERS = [
    torch.nn.Linear(INPUT_DIM, 20),
    torch.nn.LeakyReLU(0.2),
    torch.nn.Linear(20, 15),
    torch.nn.LeakyReLU(0.2),
    torch.nn.Linear(15, OUTPUT_DIM),
    torch.nn.Sigmoid()
]

MODEL = TorchPerceptronModel(INPUT_DIM, OUTPUT_DIM, LAYERS)

ALGORITHM = TorchNNAlgorithm(MODEL)
```

```python
# Now we make the algorithm
ALGORITHM.train(DATASET, epochs=500, early_stopping=True)
```
